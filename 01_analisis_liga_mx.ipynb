```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

partidos = {
    "Equipo": ["Chivas", "Cruz Azul", "America", "Atlas",
               "Toluca", "Monterrey", "León", "Pachuca"],
    "Partidos": [17, 17, 17, 17, 17, 17, 17, 17],
    "Victorias": [9, 10, 8, 8, 7, 8, 7, 7],
    "Empates": [6, 2, 6, 4, 6, 3, 5, 4],
    "Derrotas": [2, 5, 3, 5, 4, 6, 5, 6],
    "Goles_favor": [27, 37, 38, 30, 35, 29, 25, 31],
    "Goles_contra": [15, 27, 24, 33, 23, 25, 22, 20]
}

df = pd.DataFrame(partidos)
```

```python
# Puntos
df["Puntos"] = df["Victorias"] * 3 + df["Empates"]

# Diferencia de goles
df["Diferencia"] = df["Goles_favor"] - df["Goles_contra"]

# Promedio de goles anotados
df["PromGoles"] = (
    df["Goles_favor"] / df["Partidos"]
).round(2)

# Promedio de goles recibidos
df["PromGolesContra"] = (
    df["Goles_contra"] / df["Partidos"]
).round(2)

# Porcentaje de victorias
df["PorcVictorias"] = (
    df["Victorias"] / df["Partidos"] * 100
).round(2)

display(df)
```

```python
df.plot(
    x="Equipo",
    y="PromGoles",
    kind="bar",
    legend=False,
    title="Promedio de goles anotados"
)

plt.ylabel("Goles por partido")
plt.xlabel("Equipo")
plt.tight_layout()
plt.show()
```

```python
df.plot(
    x="Equipo",
    y="PromGolesContra",
    kind="bar",
    legend=False,
    title="Promedio de goles recibidos"
)

plt.ylabel("Goles por partido")
plt.xlabel("Equipo")
plt.tight_layout()
plt.show()
```

```python
# 1. Max sobre una columna
max_A = df['PromGoles'].max()
print(max_A)
```

```python
# Mínimo promedio de goles recibidos
min_A = df["PromGolesContra"].min()
print(min_A)
```

```python
equipo = df.loc[
    df["PromGolesContra"].idxmin(),
    "Equipo"
]

print(equipo)
```

```python
# Ordenar equipos por puntos
df_ordenado = df.sort_values(
    by="Puntos",
    ascending=False
)

display(df_ordenado)
```

```python
# Relación entre goles anotados y puntos

plt.scatter(
    df["Goles_favor"],
    df["Puntos"]
)

plt.title("Goles anotados vs puntos")
plt.xlabel("Goles anotados")
plt.ylabel("Puntos")

plt.grid(True)
plt.show()
```

```python
# Relación entre diferentes estadísticas y puntos

corr_goles = df["Goles_favor"].corr(df["Puntos"])
corr_contra = df["Goles_contra"].corr(df["Puntos"])
corr_diferencia = df["Diferencia"].corr(df["Puntos"])

print("Goles anotados:", round(corr_goles, 3))
print("Goles recibidos:", round(corr_contra, 3))
print("Diferencia de goles:", round(corr_diferencia, 3))
```
